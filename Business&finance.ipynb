```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression, LogisticRegression
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline

# Let's engineer the requested business metrics into our dataframe.
# Since we only have 'salary' as a financial metric in this employee dataset, we will calculate simulated aggregate business variables for analysis.
# Total Revenue = Sum of simulated Sales (using Salary as a proxy base)
np.random.seed(42)
df['Total_Sales'] = df['salary'] * np.random.uniform(1.2, 1.8, len(df))
df['Total_Expenses'] = df['salary'] * np.random.uniform(0.6, 0.9, len(df))
df['Total_Revenue'] = df['Total_Sales'] # Revenue corresponds directly to Sales here
df['Net_Profit'] = df['Total_Revenue'] - df['Total_Expenses']
df['Profit_Margin'] = df['Net_Profit'] / df['Total_Revenue']

# Display computed metrics
display(df[['salary', 'Total_Sales', 'Total_Expenses', 'Total_Revenue', 'Net_Profit', 'Profit_Margin']].head())
```

```python
# Preprocessing pipeline for modeling
features = ['age', 'city', 'remote_work', 'salary', 'performance_rating']
# Target for Linear Regression
target_reg = 'Net_Profit'
# Target for Logistic Regression (let's classify if performance_rating is high, i.e., >= 4)
df['High_Performance'] = (df['performance_rating'] >= 4).astype(int)
target_clf = 'High_Performance'

# Split features
X_data = df[features].copy()
y_reg = df[target_reg].copy()
y_clf = df[target_clf].copy()

# Handle missing values and categorical features
num_cols = ['age', 'salary', 'performance_rating']
cat_cols = ['city', 'remote_work']

preprocessor = ColumnTransformer(
    transformers=[
        ('num', SimpleImputer(strategy='median'), num_cols),
        ('cat', Pipeline([
            ('impute', SimpleImputer(strategy='most_frequent')),
            ('onehot', OneHotEncoder(handle_unknown='ignore', sparse_output=False))
        ]), cat_cols)
    ]
)

X_processed = preprocessor.fit_transform(X_data)
print("Features preprocessed successfully!")
```

```python
# Fix: Remove rows with NaN in the target y_reg (Net_Profit)
valid_indices = ~np.isnan(y_reg)
X_processed_clean = X_processed[valid_indices]
y_reg_clean = y_reg[valid_indices]

# 1. Train Linear Regression Model (to predict Net Profit)
X_train_r, X_test_r, y_train_r, y_test_r = train_test_split(X_processed_clean, y_reg_clean, test_size=0.2, random_state=42)
lin_reg = LinearRegression()
lin_reg.fit(X_train_r, y_train_r)
print(f"Linear Regression Model R^2 Score: {lin_reg.score(X_test_r, y_test_r):.4f}")

# Fix: Remove rows with NaN in target y_clf if any exist
valid_clf_indices = ~np.isnan(y_clf)
X_processed_clf = X_processed[valid_clf_indices]
y_clf_clean = y_clf[valid_clf_indices]

# 2. Train Logistic Regression Model (to classify high performance)
X_train_c, X_test_c, y_train_c, y_test_c = train_test_split(X_processed_clf, y_clf_clean, test_size=0.2, random_state=42)
log_reg = LogisticRegression(max_iter=1000)
log_reg.fit(X_train_c, y_train_c)
print(f"Logistic Regression Model Accuracy: {log_reg.score(X_test_c, y_test_c):.4f}")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import mean_absolute_error, mean_squared_error, accuracy_score, classification_report

# 1. Evaluate Linear Regression
y_pred_r = lin_reg.predict(X_test_r)
mae = mean_absolute_error(y_test_r, y_pred_r)
mse = mean_squared_error(y_test_r, y_pred_r)

print("=== LINEAR REGRESSION REPORT ===")
print(f"Mean Absolute Error (MAE): ${mae:,.2f}")
print(f"Mean Squared Error (MSE): ${mse:,.2f}")

# 2. Evaluate Logistic Regression
y_pred_c = log_reg.predict(X_test_c)
acc = accuracy_score(y_test_c, y_pred_c)

print("\n=== LOGISTIC REGRESSION REPORT ===")
print(f"Classification Accuracy: {acc * 100:.2f}%")
print("\nClassification Report:")
print(classification_report(y_test_c, y_pred_c))

# 3. Visualization of Business Metrics & Regression
plt.figure(figsize=(14, 5))

# Subplot 1: Actual vs Predicted Net Profit
plt.subplot(1, 2, 1)
sns.scatterplot(x=y_test_r, y=y_pred_r, color='teal', alpha=0.7)
plt.plot([y_test_r.min(), y_test_r.max()], [y_test_r.min(), y_test_r.max()], 'k--', lw=2)
plt.title('Actual vs Predicted Net Profit')
plt.xlabel('Actual Profit ($)')
plt.ylabel('Predicted Profit ($)')
plt.grid(True)

# Subplot 2: Profit Margin Distribution across Departments
plt.subplot(1, 2, 2)
sns.boxplot(data=df, x='department', y='Profit_Margin', palette='Set2')
plt.title('Profit Margin Distribution by Department Code')
plt.xlabel('Department Code (Encoded)')
plt.ylabel('Profit Margin')
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
import gradio as gr
import pandas as pd
import numpy as np

# Prepare functions for our Gradio interface
def get_employee_insights(emp_name):
    # Search for employee records
    match = df[df['emp_name'].str.lower() == emp_name.strip().lower()]
    if match.empty:
        return f"❌ Employee '{emp_name}' not found. Please try names like 'Employee_1', 'Employee_2', etc.", None

    row = match.iloc[0]

    # Generate formatted HTML details
    html_output = f"""
    <div style='padding: 15px; border-radius: 8px; background-color: #f7f9fc; border-left: 5px solid #2a9d8f;'>
        <h3>👤 Employee Profile: {row['emp_name']}</h3>
        <p><b>Age:</b> {row['age'] if not pd.isna(row['age']) else 'N/A'}</p>
        <p><b>City:</b> {row['city']}</p>
        <p><b>Remote Work:</b> {row['remote_work']}</p>
        <p><b>Performance Rating:</b> {row['performance_rating'] if not pd.isna(row['performance_rating']) else 'N/A'}/5</p>
        <hr style='border: 0.5px solid #ccc;' />
        <h4>📊 Financial Indicators:</h4>
        <p><b>Salary:</b> ${row['salary']:,.2f if not pd.isna(row['salary']) else 0.00}</p>
        <p><b>Total Sales Generated:</b> ${row['Total_Sales']:,.2f if not pd.isna(row['Total_Sales']) else 0.00}</p>
        <p><b>Total Operating Expenses:</b> ${row['Total_Expenses']:,.2f if not pd.isna(row['Total_Expenses']) else 0.00}</p>
        <p><b>Net Profit Contribution:</b> <span style='color: green; font-weight: bold;'>${row['Net_Profit']:,.2f if not pd.isna(row['Net_Profit']) else 0.00}</span></p>
        <p><b>Computed Profit Margin:</b> {row['Profit_Margin']*100:.2f if not pd.isna(row['Profit_Margin']) else 0.00}%</p>
    </div>
    """

    # Summary statistics indicator data
    metrics = {
        "Total Revenue Contribution": f"${row['Total_Revenue']:,.2f}" if not pd.isna(row['Total_Revenue']) else "$0.00",
        "Net Margin Efficiency": f"{row['Profit_Margin']*100:.1f}%" if not pd.isna(row['Profit_Margin']) else "0.0%",
    }
    return html_output, pd.DataFrame([metrics])

# Define the Gradio Interface layout
with gr.Blocks(theme=gr.themes.Soft()) as demo:
    gr.Markdown("# 📈 Employee Enterprise & Business Metrics Dashboard")
    gr.Markdown("Analyze key metrics like Total Sales, Revenue, Expenses, Net Profit, and run ML-driven insights on your imported data.")

    with gr.Row():
        # Left Column for query controls
        with gr.Column(scale=1):
            emp_input = gr.Textbox(
                label="Enter Employee Name (e.g., Employee_1, Employee_5)",
                placeholder="Employee_1",
                value="Employee_1"
            )
            btn = gr.Button("🔍 Generate Profile & Insights", variant="primary")

        # Right Column for profiles and KPIs
        with gr.Column(scale=2):
            profile_html = gr.HTML(label="Employee Profile Details")
            metrics_table = gr.Dataframe(label="KPI Overview")

    # Bind button action
    btn.click(fn=get_employee_insights, inputs=emp_input, outputs=[profile_html, metrics_table])

# Launch dashboard locally inside Google Colab
demo.launch(share=True, debug=False)
```
